```python
import os
import django

os.environ.setdefault("DJANGO_SETTINGS_MODULE", "config.settings")
os.environ["DJANGO_ALLOW_ASYNC_UNSAFE"] = "true"
django.setup()

from asgiref.sync import sync_to_async
import pandas as pd
from nodepoint.models import Workspace, KnowledgeEntity, KnowledgeRelation, Document
```

```python
data = []

for ent in KnowledgeEntity.objects.all():
    ws = ent.document.workspace
    data.append({
        "ws_id": ws.pk,
        "ent_name": ent.name,
        "ent_type": ent.entity_type,
        "attr": "".join(f"{key}: {value} "for key, value in ent.attributes.items())
    })
```

```python
df = pd.DataFrame(data)
df.head(20)
```
