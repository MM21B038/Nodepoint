```python
url = "http://10.10.112.17:8000/planner/"
```

```python
prompts = []
```

```python
import requests
```

```python
for i in range(1, 1025):
    response = requests.get(f"{url}{i}")

    if response.ok:
        data = response.json()
        prompt = data.get("prompt")

        if prompt:
            prompts.append((i, prompt))
```
